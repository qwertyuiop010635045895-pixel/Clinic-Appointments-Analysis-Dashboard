```python
import pandas as pd
Clinic_Appointments= pd.read_csv(r"C:\Users\qwert\OneDrive\Desktop\New folder\Project\Clinic_Appointments\DATA\Clinic_Appointments.csv")

# قائمة أسماء ذكور وإناث
male_names = [
    "Ahmed", "Mohamed", "Mahmoud", "Omar", "Ali",
    "Youssef", "Mostafa", "Khaled", "Hassan", "Ibrahim"
]

female_names = [
    "Mona", "Sara", "Fatma", "Amina", "Nour",
    "Aya", "Mariam", "Yara", "Nada", "Heba"
]

def get_gender(name):
    first_name = str(name).split()[0].title()

    if first_name in male_names:
        return "Male"
    elif first_name in female_names:
        return "Female"
    else:
        return None  # أو "Unknown"

# تحديث العمود
Clinic_Appointments["Gender"] = Clinic_Appointments["PatientName"].apply(get_gender)
```
